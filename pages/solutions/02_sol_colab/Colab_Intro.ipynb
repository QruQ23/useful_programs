```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import base64
import os

# Parametry fizyczne wahadła
g = 9.81  # przyspieszenie ziemskie (m/s^2)
L = 1.0   # długość wahadła (m)
theta0 = np.radians(45)  # kąt początkowy (45 stopni przekonwertowane na radiany)

# Czas symulacji
t_max = 10.0
fps = 30
dt = 1.0 / fps
t = np.arange(0, t_max, dt)

# Rozwiązanie przybliżone dla małych drgań (analityczne)
theta = theta0 * np.cos(np.sqrt(g / L) * t)

# Współrzędne końca wahadła w układzie kartezjańskim
x_pendulum = L * np.sin(theta)
y_pendulum = -L * np.cos(theta)

# Tworzenie figury i osi wykresu
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.2 * L, 1.2 * L)
ax.set_ylim(-1.2 * L, 0.2 * L)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title("Animacja Wahadła Matematycznego")

# Elementy graficzne animacji
line, = ax.plot([], [], 'o-', lw=2, color='darkblue', markersize=10)
trace, = ax.plot([], [], '-', lw=1, color='lightblue', alpha=0.6)  # Ślad ruchu

def init():
    line.set_data([], [])
    trace.set_data([], [])
    return line, trace

def animate(i):
    # Aktualna pozycja kulki
    this_x = [0, x_pendulum[i]]
    this_y = [0, y_pendulum[i]]
    line.set_data(this_x, this_y)

    # Rysowanie śladu ruchu z poprzednich klatek
    trace.set_data(x_pendulum[:i], y_pendulum[:i])
    return line, trace

# Tworzenie obiektu animacji
ani = animation.FuncAnimation(
    fig, animate, init_func=init, frames=len(t), interval=dt*1000, blit=True
)

# Zamykamy figurę, aby zapobiec wyświetleniu jej jako statycznego obrazu
plt.close(fig)

# Zapisywanie animacji do pliku video.mp4 za pomocą ffmpeg
video_path = "video.mp4"
ani.save(video_path, writer='ffmpeg', fps=fps)

# Wyświetlanie (osadzanie) wideo w notebooku Colab przy użyciu odtwarzacza HTML5
print(f"Plik wideo został pomyślnie zapisany w lokalizacji: {os.path.abspath(video_path)}")

mp4 = open(video_path, 'rb').read()
data_url = "data:video/mp4;base64," + base64.b64encode(mp4).decode()
HTML(f"""
<video width="500" height="500" controls>
      <source src="{data_url}" type="video/mp4">
</video>
""")
```

```python
# Correcting the label for rendering issue in matplotlib mathtext
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import os

# Parametric equations
t = np.linspace(-20, 3 * np.pi, 5000)
x = np.sin(t) * np.cos(t)
y = np.tan(t) * np.sin(t)
z = np.tan(t) * np.cos (t)

# 3D Plot
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="Parametric Curve")
ax.set_title("3D Parametric Curve", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# Zapisanie wykresu do pliku parametric_curve.png
output_path = "parametric_curve.png"
plt.savefig(output_path, dpi=300, bbox_inches='tight')
print(f"Wykres został pomyślnie zapisany w lokalizacji: {os.path.abspath(output_path)}")

# Wyświetlenie wykresu
plt.show()
```
